```python
x = set(all_govs_df["GOV_NAME_A"].values)
```

```python
x = list(map(lambda x : x.replace("محافظه", "").strip(), x ))
x = list(map(lambda x : x.replace("محافظة", "").strip(), x ))

x = list(map(lambda x : x.replace("المجلس الأعلى لمدينة", "").strip(), x ))
x = list(map(lambda x : x.replace("\xa0", " ").strip(), x ))


x = x + list(map(lambda x : x.replace("ال", " ").strip(), x ))
```

```python
x
```

```python
all_govs_df = all_govs_df.drop(columns=["Unnamed: 0", "area"])
# First fixing
# Remove all punctuation

def remove_punctuation(x):
  x["name"] = re.sub(r'[^\w\s]','',str(x["name"]))
  x["SSEC_NAM_1"] = re.sub(r'[^\w\s]','',str(x["SSEC_NAM_1"]))
  x["SEC_NAME_A"] = re.sub(r'[^\w\s]','',str(x["SEC_NAME_A"]))
  x["GOV_NAME_A"] = re.sub(r'[^\w\s]','',str(x["GOV_NAME_A"]))

  # Remove محافظة
  x["GOV_NAME_A"] = x["GOV_NAME_A"].replace("محافظه", "").strip()
  x["GOV_NAME_A"] = x["GOV_NAME_A"].replace("محافظة", "").strip()
  x["GOV_NAME_A"] = x["GOV_NAME_A"].replace("المجلس الأعلى لمدينة", "").strip()


  return x

all_govs_df = all_govs_df.apply(remove_punctuation, axis=1)
```

```python
all_govs_df.to_csv("govs_streets.csv")
```
