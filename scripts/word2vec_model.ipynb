```python
# Load your text file
with open('street_names.txt', 'r', encoding='utf-8') as file:
    sentences = [line.split() for line in file.readlines()]
```

```python
import gensim
from gensim.models import Word2Vec



# Create Word2Vec model
model = Word2Vec(sentences, vector_size=256, window=5, min_count=1, workers=4, epochs=100)

# Save the model for later use
# model.save("word2vec_model.model")
```

```python
# Example sentence
sentence = ["الغربية", "طنطا", "رضوان", "حسن", "شارع"]
context_words = [word for word in sentence if word in model.wv.key_to_index]
```

```python
x = "شارع مكتب العمل ميت غمر الاهلي"
sentence = x.split()


nom = len(sentence) - 1 

for i, word in enumerate(sentence):
    distance = 0.0
    for j, word2 in enumerate(sentence):
        if i == j : continue
        distance += float(np.dot(model.wv[word], model.wv[word2]))
    score = distance / nom
    print(word, score)
        
        

```

```python
import nltk
from nltk.tokenize import word_tokenize
from collections import defaultdict

# Read the text file
with open('street_names.txt', 'r', encoding='utf-8') as file:
    text = file.read()

# Download necessary NLTK data files (run this once)
nltk.download('punkt')

tokens = word_tokenize(text)

# Create a mapping of words to unique IDs
word_to_id = {}
id_to_word = {}

for idx, word in enumerate(set(tokens)):
    word_to_id[word] = idx
    id_to_word[idx] = word

# Display the mappings
print("Word to ID mapping:")
for word, id in word_to_id.items():
    print(f"{word}: {id}")

print("\nID to Word mapping:")
for id, word in id_to_word.items():
    print(f"{id}: {word}")
```

```python
from collections import Counter
from collections import defaultdict

def get_forward_backward_probabilites(filename):
    forward_occurance = []
    backward_occurance = []
    word_occurance = []

    with open (filename, "r") as file:
        for line in file.readlines():
            words_list =line.split(" ") 
            words_list = list(map(lambda x: x.strip(), words_list))
            for i, word in enumerate(words_list):
                word_occurance.append(word)
                for j in range(i+1, len(words_list)):
                    forward_occurance.append( (word, words_list[j]) )
                    backward_occurance.append( (words_list[j], word) )



    forward_frequncies = Counter(forward_occurance)
    backward_frequncies = Counter(backward_occurance)
    word_frequncies = Counter(word_occurance)

    forward_probabilities = defaultdict(float)
    backward_probabilities = defaultdict(float)


    for (w1, w2), count in forward_frequncies.items():
        forward_probabilities[(w1, w2)] = count / word_frequncies[w1]

    for (w1, w2), count in backward_frequncies.items():
        backward_probabilities[(w1, w2)] = count / word_frequncies[w1]

    return forward_probabilities, backward_probabilities
```
